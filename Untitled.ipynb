```python
import torch
import numpy as np
from PIL import Image
from transformers import CLIPProcessor, CLIPModel
```

```python
model_name = "openai/clip-vit-base-patch32"

processor = CLIPProcessor.from_pretrained(model_name)
model = CLIPModel.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)
model.eval()

print("CLIP Model Loaded Successfully")
print("Device:", device)
```

```python
def image_to_vector(image_path):

    image = Image.open(image_path).convert("RGB")

    inputs = processor(
        images=image,
        return_tensors="pt"
    )

    inputs = {key: value.to(device) for key, value in inputs.items()}

    with torch.no_grad():

        image_features = model.get_image_features(**inputs)

        # Handle different Transformers output formats
        if hasattr(image_features, "pooler_output"):
            image_features = image_features.pooler_output

        elif hasattr(image_features, "last_hidden_state"):
            image_features = image_features.last_hidden_state[:, 0, :]

        # Normalize vector
        image_features = torch.nn.functional.normalize(
            image_features, p=2, dim=-1
        )

    return image_features.cpu().numpy().flatten()
```

```python
query_vector = image_to_vector(train_paths[0])

print("Vector Shape:", query_vector.shape)
print("Vector Type:", type(query_vector))
```

```python
train_embeddings = []

for path in train_paths:

    vector = image_to_vector(path)

    train_embeddings.append(vector)

train_embeddings = np.array(train_embeddings)

print("Embedding Shape:", train_embeddings.shape)
```

```python
query_image = test_paths[0]

query_vector = image_to_vector(query_image)

print("Query Vector Shape:", query_vector.shape)
```

```python
similarities = np.dot(train_embeddings, query_vector)

print(similarities)
```

```python
top_indices = np.argsort(similarities)[::-1][:5]

for index in top_indices:
    print(
        "Image:", train_paths[index],
        "| Similarity:", similarities[index]
    )
```

```python
plt.figure(figsize=(15, 5))

for i, index in enumerate(top_indices):

    img = Image.open(train_paths[index]).convert("RGB")

    plt.subplot(1, 5, i + 1)
    plt.imshow(img)

    plt.title(f"Similarity: {similarities[index]:.3f}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import faiss
import numpy as np
import os
import json

print("FAISS version:", faiss.__version__)
```

```python
os.makedirs("artifacts", exist_ok=True)

print("Artifacts folder created successfully")
```

```python
print("Embeddings Shape:", train_embeddings.shape)
print("Data Type:", train_embeddings.dtype)
```

```python
np.save("artifacts/ring_embeddings.npy", train_embeddings)

print("Embeddings saved successfully!")
```

```python
# Convert embeddings to float32
embeddings = np.ascontiguousarray(
    train_embeddings,
    dtype=np.float32
)

# Get embedding dimensions
dimension = embeddings.shape[1]

# Create FAISS index
index = faiss.IndexFlatIP(dimension)

# Add embeddings
index.add(embeddings)

print("FAISS Index Created Successfully!")
print("Total vectors:", index.ntotal)
print("Dimensions:", index.d)
```

```python
faiss.write_index(
    index,
    "artifacts/ring_faiss.index"
)

print("FAISS Index Saved Successfully!")
```

```python
with open("artifacts/image_paths.json", "w") as f:
    json.dump(train_paths, f)

print("Image paths saved successfully!")
```

```python
# Load embeddings
loaded_embeddings = np.load(
    "artifacts/ring_embeddings.npy"
)

# Load FAISS index
loaded_index = faiss.read_index(
    "artifacts/ring_faiss.index"
)

# Load image paths
with open("artifacts/image_paths.json", "r") as f:
    loaded_paths = json.load(f)


print("Embeddings:", loaded_embeddings.shape)
print("FAISS vectors:", loaded_index.ntotal)
print("Image paths:", len(loaded_paths))
```

```python
# Select query image
query_image = test_paths[0]

# Convert query image into vector
query_vector = image_to_vector(query_image)

# Convert into FAISS-compatible format
query_vector = np.ascontiguousarray(
    query_vector.reshape(1, -1),
    dtype=np.float32
)

# Search Top 5 similar images
distances, indices = loaded_index.search(
    query_vector, 5
)

print("Similarity Scores:", distances)
print("Image IDs:", indices)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

plt.figure(figsize=(15, 5))

for i, idx in enumerate(indices[0]):

    img_path = loaded_paths[idx]

    img = Image.open(img_path).convert("RGB")

    plt.subplot(1, 5, i + 1)
    plt.imshow(img)
    plt.title(f"Similarity: {distances[0][i]:.3f}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def search_similar_images(query_path, top_k=5):

    # Convert query image into vector
    query_vector = image_to_vector(query_path)

    query_vector = np.ascontiguousarray(
        query_vector.reshape(1, -1),
        dtype=np.float32
    )

    # Search using FAISS
    distances, indices = loaded_index.search(
        query_vector, top_k
    )

    return distances[0], indices[0]
```

```python
for query_path in test_paths[:5]:

    distances, indices = search_similar_images(query_path)

    print("Query Image:", os.path.basename(query_path))
    print("Similarity Scores:", distances)
    print("Retrieved Images:")

    for idx in indices:
        print(loaded_paths[idx])

    print("-" * 60)
```

```python
def display_search_results(query_path, top_k=5):

    distances, indices = search_similar_images(
        query_path, top_k
    )

    plt.figure(figsize=(16, 5))

    # Query image
    plt.subplot(1, top_k + 1, 1)

    plt.imshow(Image.open(query_path).convert("RGB"))
    plt.title("Query Image")
    plt.axis("off")

    # Similar images
    for i, idx in enumerate(indices):

        img_path = loaded_paths[idx]

        img = Image.open(img_path).convert("RGB")

        plt.subplot(1, top_k + 1, i + 2)
        plt.imshow(img)

        plt.title(f"Score: {distances[i]:.3f}")
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
for query_path in test_paths[:5]:

    display_search_results(query_path)
```
